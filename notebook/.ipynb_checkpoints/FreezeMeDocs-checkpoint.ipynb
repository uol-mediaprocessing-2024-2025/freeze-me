In the following we will go through all steps of the project, their development history and other things worth mentioning (Future Improvements, technical difficulties, etc.).

```python
from concurrent.futures import ThreadPoolExecutor

def read_image(file_path):
    return cv2.imread(file_path, cv2.IMREAD_UNCHANGED)

def load_video_frames(paths):
    with ThreadPoolExecutor() as executor:
        images = list(executor.map(read_image, paths))
    return images
```

### 2.2 Getting Video Details

#### 2.2.1 Current Process

```python
async def get_video_details(video_id):
    try:
        path = get_upload_path(video_id)
        image_folder = get_images_path(video_id)
        details = ffmpeg.probe(path.__str__(), cmd="static_ffprobe")
        video_stream = None
        for stream in details["streams"]:
            if stream["codec_type"] == "video":
                video_stream = stream
                break
        global fps
        total_frames = len(os.listdir(image_folder))
        details["total_frames"] = total_frames
        fps_string = video_stream["r_frame_rate"]
        slash = fps_string.find("/")
        fps = round(float(fps_string[0:slash]) / float(fps_string[slash + 1:]), 2)
        print("FPS: ", fps)
        print("Total frames: ", total_frames)
        return details
    except Exception as e:
        print(e)
        print(e.__traceback__)
        print(traceback.format_exc())

        return ""
```

### 2.3 Cutting video

#### 2.3.1 Current Process

```python
import os
import ffmpeg
import shutil
from IPython.display import Video

fps = 60  

def cut_video(video_path: str, output_path: str, start_time: float, end_time: float):
    """
    Cuts a video from start_time to end_time

    :param video_path: path to original video
    :param output_path: path to cut video
    :param start_time: starttime in seconds
    :param end_time: endtime in seconds
    """
    
    print(f"Video Path: {video_path}")
    print(f"Output Path: {output_path}")


    try:
        if not os.path.exists(video_path):
            raise FileNotFoundError(f"Originalvideo {video_path} wurde nicht gefunden.")

        start_frame = int(start_time * fps)
        end_frame = int(end_time * fps)

        print(f"Schneide Video von {start_time}s ({start_frame} Frames) bis {end_time}s ({end_frame} Frames)")

        # Cut video using ffmpeg
        input_file = ffmpeg.input(video_path)

        ffmpeg.output(
            input_file.trim(start_frame=start_frame, end_frame=end_frame).setpts('PTS-STARTPTS'),
            output_path, vcodec='libx264', movflags='faststart', an=None
        ).overwrite_output().run(quiet=True)

        if os.path.exists(output_path):
            print(f"Geschnittenes Video gespeichert unter: {output_path}")
            return output_path
        else:
            raise Exception(f"Das geschnittene Video wurde nicht erfolgreich erstellt.")

    except Exception as e:
        print(f"Fehler beim Schneiden des Videos: {e}")
        return None


# Cut the video
cut_video(str(get_sample_video_path()), str(get_cut_video_path()), start_time=11, end_time=15)

# Display the resulting video
Video(get_cut_video_path(), embed=True, html_attributes="controls muted autoplay", width=640, height=360)
```

### 3.2 Segmenting video

#### 3.2.1 Current Process

```python
async def initialize_segmentation():
    try:
        global inference_state, points, labels
        image_folder = get_sample_frames_path()     # in project a separate folder for each project is used
        total_frames = len(os.listdir(image_folder))
        points = [[]] * total_frames
        labels = [[]] * total_frames
        print(points)
        print(labels)
        inference_state = predictor.init_state(video_path=image_folder.__str__())
    except Exception as e:
        print(e)
        print(e.__traceback__)
```

```python
import cv2
import supervision as sv
import numpy as np

colors = ['#FF1493', '#00BFFF', '#FF6347', '#FFD700']
mask_annotator = sv.MaskAnnotator(
    color=sv.ColorPalette.from_hex(colors),
    color_lookup=sv.ColorLookup.TRACK)

async def add_new_point_to_segmentation(point_x, point_y, point_type, frame_num):
    try:
        global points, labels
        points[frame_num].append([point_x, point_y])
        labels[frame_num].append(point_type)
        print(points, labels)
        _, out_obj_ids, out_mask_logits = predictor.add_new_points_or_box(
            inference_state=inference_state,
            frame_idx=frame_num,
            obj_id=1,
            points=points[frame_num],
            labels=labels[frame_num],
        )
        mask_data = (out_mask_logits > 0.0).cpu().numpy()
        n, x, h, w = mask_data.shape
        masks = mask_data.reshape(n * x, h, w)
        detections = sv.Detections(
            xyxy=sv.mask_to_xyxy(masks=masks),
            mask=masks,
            tracker_id=np.array(out_obj_ids)
        )
        frame_path = get_frame_path(frame_num)
        frame = cv2.imread(frame_path.__str__())
        frame = mask_annotator.annotate(frame, detections)
        cv2.imshow("Point selected frame", frame)
    except Exception as e:
        print(e)
        print(e.__traceback__)
```

In the following get_masked_video()-Method we use SAM2.1 and Supervision to separate the traced object from its background.  <br>
The separated fore- and background are then saved, as well as a video with a mask showing the results is created.  <br>
In the project the frames are saved on disk, while in this notebook they are just stored in an array.

```python
import ffmpeg

async def get_masked_video():
    try:
        output_path = get_masked_video_path()
        image_path = get_sample_frames_path()
        video_path = get_sample_video_path()
        video_info = sv.VideoInfo.from_video_path(video_path.__str__())
        frames_paths = sorted(sv.list_files_with_extensions(directory=image_path.__str__(), extensions=["jpeg"]))
        background_frames = []
        foreground_frames = []
        # run propagation throughout the video and collect the results in a dict
        temp_file = get_temp_file_path()
        temp_file.mkdir(parents=True, exist_ok=True)
        frames = load_video_frames(frames_paths)
        with sv.VideoSink(temp_file.__str__(), video_info=video_info) as sink:
            for out_frame_idx, out_obj_ids, out_mask_logits in predictor.propagate_in_video(inference_state, start_frame_idx=0):
                frame = frames[out_frame_idx]
                masks = (out_mask_logits > 0.0).cpu().numpy()
                n, x, h, w = masks.shape
                masks = masks.reshape(n * x, h, w)
                detections = sv.Detections(
                    xyxy=sv.mask_to_xyxy(masks=masks),
                    mask=masks,
                    tracker_id=np.array(out_obj_ids)
                )

                # Create foreground cut frames
                transparent_foreground = cv2.cvtColor(frame, cv2.COLOR_BGR2BGRA)
                transparent_foreground[:, :, 3] = masks.astype(np.uint8) * 255
                foreground_frames.append(transparent_foreground)

                # Create background cut frames
                transparent_background = cv2.cvtColor(frame, cv2.COLOR_BGR2BGRA)
                transparent_background[:, :, 3] = ((masks.astype(np.uint8) + 1) % 2) * 255
                background_frames.append(transparent_background)

                # Create masked frames
                frame = mask_annotator.annotate(frame, detections)
                sink.write_frame(frame)

        (ffmpeg
         .input(temp_file.__str__())
         .output(
            output_path.__str__(),
            vcodec='libx264',
            movflags='faststart',
            an=None
        )
         .overwrite_output().run(quiet=True)
         )
        temp_file.unlink(missing_ok=True)
        print(output_path.__str__())
        return output_path.__str__()
    except Exception as e:
        print(e)
        print(e.__traceback__)
```
